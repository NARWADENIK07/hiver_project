finding the root of tweet

```python
# Dictionary to store roots that we have already found
# Example: {6: 8, 5: 8, 7: 8}
root_cache = {}
```

```python
# Dictionary used to remember roots we have already calculated
root_cache = {}


def find_root(tweet_id):
    # If we already know the root, return it immediately
    if tweet_id in root_cache:
        return root_cache[tweet_id]

    # Start from the current tweet
    current_id = tweet_id

    # Keep following the parent until there is no parent
    while pd.notna(current_id):

        # Find the parent of the current tweet
        parent_id = parent_map.get(current_id)

        # If there is no parent, this tweet is the root
        if pd.isna(parent_id):
            root_cache[tweet_id] = current_id
            return current_id

        # Move to the parent tweet
        current_id = int(parent_id)

    # Return None if a root cannot be found
    return None
```

```python
# Test the optimized root-finding function
print(find_root(5))
print(find_root(6))
print(find_root(7))

# Check what has been stored in the cache
print(root_cache)
```
